# 🛠 Tools (not published)

```python
# Word count — counts published review prose only (untagged markdown, headings & placeholders excluded)
import re, nbformat

nb_ = nbformat.read("review.ipynb", as_version=4)
SKIP = {"prompt", "exclude", "header", "figure", "references"}
sections, current = {}, "(before first heading)"
for c in nb_.cells:
    if c.cell_type != "markdown" or SKIP & set(c.metadata.get("tags", [])):
        continue
    for line in c.source.splitlines():
        if line.startswith("#"):
            current = line.lstrip("# ").strip()
            sections.setdefault(current, 0)
            continue
        line = re.sub(r"^\*\[.*\]\*$", "", line.strip())  # unfilled placeholders
        line = re.sub(r"!\[.*?\]\(.*?\)|\[([^\]]*)\]\([^)]*\)", r"\1", line)
        sections[current] = sections.get(current, 0) + len(re.findall(r"[A-Za-z0-9’'-]+", line))

total = sum(sections.values())
for name, n in sections.items():
    print(f"{name:<32} {n:>4}")
print(f"{'TOTAL':<32} {total:>4}   (target 700–800)  {'✅' if 700 <= total <= 800 else '⚠️'}")
```
